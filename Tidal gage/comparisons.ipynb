## Import statements and global variables

```python
import pandas as pd
import numpy as np
from datetime import datetime
import matplotlib.pyplot as plt
import seaborn as sns

START_TIME = datetime(2022, 7, 14, 18, 0, 0, 0)
END_TIME = datetime(2022, 7, 15, 6, 0, 0, 0)
```

## Befus et al. model

```python
befus_df = pd.read_csv('hr_heb_Verified_(m)_38834_dt360s_230418.csv')
befus_df['date_time'] = pd.to_datetime(befus_df['date_time'])
befus_df = befus_df[befus_df['date_time'].between(START_TIME, END_TIME)]
display(befus_df)
```

```python
sns.lineplot(x='date_time', y='Qtotal_m3s', data=befus_df)
plt.xlabel('Time')
plt.ylabel(r'Discharge ($m^3/s$)')
plt.axhline(0, color='black', linestyle=':')
```
